```python
n = int(input("Enter num: "))

if n%5==0 and n%11==0:
    print("Divisible by both 5 and 11")
elif n%5==0:
    print("Divisible by only 5")
elif n%11==0:
    print("Divisible by only 11")
else:
    print("Not divisible by both")
```
