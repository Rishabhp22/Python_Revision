```python
n = int(input("Enter num: "))
rev = 0
temp_n = n

while temp_n>0:
    digit = temp_n%10
    rev = rev*10+digit
    temp_n//=10

if rev == n:
    print(f"{n} Is the Palindrome number.")
else:
    print("Not a Palindrome number")
```

```python
n = 564111222

ec = 0
oc = 0

while n>0:
    digit = n%10
    if digit%2==0:
        ec+=1
    else:
        oc+=1
    n//=10
print(ec)
print(oc)
```

```python
n = 5
a = 0
b = 1
Fib = 0

while Fib<n:
    print(a,end=" ")
    next = a+b
    a=b
    b = next
    Fib+=1
```

```python
n = 9474
c = 0

while n >0:
    digit=n%10
    c+=1
    n//=10


total = 0

while n>0:
    digit = n%10
    total+= digit**c
    n//=10

if total == n:
    print("Armstrong Num")
else:
    print("Not Armstrong Num")
```

```python
n = 145
temp_n = n
total = 0

while temp_n>0:
    digit = temp_n%10
    fact = 1
    for i in range(1,digit+1):
        fact*=i

    total+=fact
    temp_n//=10

if total == n:
    print("Strong Num")
else:
    print("Not a strong num")

```

```python
a = 12
b = 18

x = a
y = b

while y != 0:
    x, y = y, x % y

hcf = x
lcm = (a * b) // hcf

print("HCF =", hcf)
print("LCM =", lcm)
```
