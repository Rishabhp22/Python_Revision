```python
l = 6   # columns
b = 4   # rows

for i in range(b):
    for j in range(l):
        if i == 0 or i == b - 1 or j == 0 or j == l - 1:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python
n = 5

for i in range(n):

    # Print leading spaces
    for j in range(n - i - 1):
        print(" ", end="")

    # Print pyramid
    for j in range(2 * i + 1):

        if j == 0 or j == 2 * i or i == n - 1:
            print("*", end="")
        else:
            print(" ", end="")

    print()
```

```python
for num in range(2,101):

    is_prime = True

    for i in range(2,int(num**0.5)+1):
        if num%i == 0:
            is_prime= False
            break

    if is_prime:
        print(num,end=' ')
```
