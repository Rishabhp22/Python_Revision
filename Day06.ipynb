```python
l = [10,20,30,40,550]
total = 0
maxi = l[0]
mini = l[0]

for i in l:
    total+=i


    if i>maxi:
        maxi = i


    if i<mini:
        mini = i

print(total)
print(maxi)
print(mini)
```

```python
l = [10,20,30,40,550]
n = int(input())

for i in range(len(l)):
    if l[i] == n:
        print('Index = ',i)
        break
else:
    print('Element not found')
```

```python
l = [11,15,11,15,20]
n = int(input())
count = 0

for i in l:
    if i == n:
        count+=1
print('Frequency', count)
```

```python
l = [11,15,11,15,20]

for i in l:
    print(i, '--->',l.count(i))
```

```python
l = [11, 15, 11, 15, 20]
visited = []

for i in l:
    if i not in visited:
        print(i, "->", l.count(i))
        visited.append(i)
```
