```python
t = (15, 7, 29, 3, 18, 12)
maxi = t[0]
mini = t[0]

for i in t:
    if i>maxi:
        maxi=i
    if i<mini:
        mini = i

print(maxi)
print(mini)
```

```python
t = (1, 2, 3, 4, 5)
rev_t = ()

for i in t:
    rev_t = (i,)+rev_t
print(rev_t)
```

```python
t = (1, 2, 2, 3, 4, 4, 5, 2)
visited = ()

for i in t:
    if i not in visited:
        visited += (i,)
print(visited)
```

```python
t = (10, 5, 25, 18, 25, 7)
lar = float('-inf')
sec_lar = float('-inf')

for i in t:
    if i>lar:
        sec_lar = lar
        lar = i

    elif lar > i > sec_lar:
        sec_lar = i

print(lar)
print(sec_lar)
```

```python
t = (("name", "Rishabh"), ("age", 20), ("city", "Delhi"))
dic = {}

for key,value in t:
    dic[key] = value

print(dic)
```

```python
t = (2, 4, 3, 5, 7, 8, 9)
target = 11

for i in range(len(t)):
    for j in range(i+1,len(t)):
        if t[i] + t[j] == target:
            print((t[i],t[j]))
```

```python
t = (1, 2, 3, 4, 5, 6)
k = 2

k%=len(t)
new_t = ()

for i in range(k,len(t)):
    new_t+=(t[i],)
for i in range(k):
    new_t+=(t[i],)

print(new_t) 
```

```python
t = (1, (2, 3), (4, (5, 6)), 7)

total = 0

for i in t:
    if isinstance(i, tuple):
        for j in i:
            if isinstance(j, tuple):
                for k in j:
                    total += k
            else:
                total += j
    else:
        total += i

print(total)
```
