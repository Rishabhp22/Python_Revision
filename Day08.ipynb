```python
nums = [2, 7, 11, 15, 3, 6]
target = 9

visited = set()

for num in nums:
    complement = target - num

    if complement in visited:
        print(True)
        break

    visited.add(num)
else:
    print(False)
```

```python
students = {
    "Riya": {"Python", "SQL", "Excel"},
    "Aman": {"Python", "Java"},
    "Rohit": {"Python", "SQL"},
    "Neha": {"Excel", "Power BI", "SQL"}
}

# 1. All unique skills
all_skills = set()

for skills in students.values():
    all_skills.update(skills)

print("All Skills:", all_skills)

# 2. Common skills
common = None

for skills in students.values():
    if common is None:
        common = skills
    else:
        common = common.intersection(skills)

print("Common Skills:", common)

# 3. Students who know Python
print("Knows Python:")
for name, skills in students.items():
    if "Python" in skills:
        print(name)

# 4. Skills only Neha knows
others = set()

for name, skills in students.items():
    if name != "Neha":
        others.update(skills)

print("Only Neha Knows:", students["Neha"] - others)

# 5. Students who don't know SQL
print("Don't Know SQL:")
for name, skills in students.items():
    if "SQL" not in skills:
        print(name)
```

```python
data = [
    {"A", "B", "C"},
    {"B", "C", "D"},
    {"C", "D", "E"},
    {"C", "F"}
]

# 1. Common elements
common = data[0]

for s in data[1:]:
    common = common.intersection(s)

print("Common:", common)

# 2. Frequency
freq = {}

for s in data:
    for item in s:
        freq[item] = freq.get(item, 0) + 1

print("At least two sets:")
for k, v in freq.items():
    if v >= 2:
        print(k)

print("Exactly once:")
for k, v in freq.items():
    if v == 1:
        print(k)

# 4. Largest set
largest = max(data, key=len)
print("Largest Set:", largest)

# 5. Total distinct elements
all_items = set()

for s in data:
    all_items.update(s)

print("Distinct Elements:", all_items)
print("Count:", len(all_items))
```
