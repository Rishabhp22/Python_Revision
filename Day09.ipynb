```python
data = [1, 2, 3, 2, 4, 1, 5, 3, 2]

unique = []
count = {}

for item in data:
    if item not in count:
        count[item] = 1
        unique.append(item)
    else:
        count[item]+=1

print('Unique list:', unique)
print('\nocurrences:')
for key, value in count.items():
    print(key,'-->',value)
```

```python
d1 = {
    'A':5,
    'B':10,
    'C':15
}
d2 = {
    'A':5,
    'C':10,
    'D':5
}
merge = {}
for key, value in d1.items():
    if key in d2:
        merge[key] = value + d2[key]
    else:
        merge[key] = value

for key, value in d2.items():
    if key not in d1:
        merge[key] = value

print(merge)
```

```python
students = {
    "Rishabh": "A",
    "Rahul": "B",
    "Ankit": "A",
    "Priya": "C",
    "Riya": "B",
    "Aman": "A"
}

group = {}

for key, value in students.items():
    if value in group:
        group[value].append(key)
    else:
        group[value] = [key]

print(group)
```

```python
words = [
    "apple",
    "ant",
    "banana",
    "ball",
    "cat",
    "car",
    "dog"
]

group = {}

for word in words:
    first = word[0]

    if first in group:
        group[first].append(word)
    else:
        group[first] = [word]

print(group)
```

```python
s = "aabbcdde"

freq = {}

# Count frequency
for ch in s:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1

# Find first non-repeating character
for ch in s:
    if freq[ch] == 1:
        print(ch)
        break
```
