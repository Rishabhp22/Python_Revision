```python
def is_prime(n):
    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False

    i = 3
    while i * i <= n:
        if n % i == 0:
            return False
        i += 2

    return True

# Example
num = int(input("Enter a number: "))
if is_prime(num):
    print("Prime")
else:
    print("Not Prime")
```

```python
def char_freq(s):
    freq = {}

    for ch in s:
        if ch in freq:
            freq[ch]+=1
        else:
            freq[ch] = 1
    return freq

print(char_freq('hello'))
    
```

```python
def sec_lar(l):
    sec = float('-inf')
    lar = float('-inf')
    for i in l:
        if i>lar:
            sec = lar
            lar = i
        elif lar>i>sec:
            sec=i
    return sec

print(sec_lar([1,2,25,58,64]))
```

```python
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

def lcm(a, b):
    return (a * b) // gcd(a, b)

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

print("LCM =", lcm(num1, num2))
```

```python
def add(a, b):
    return a + b

def sub(a, b):
    return a - b

def mul(a, b):
    return a * b

def div(a, b):
    if b == 0:
        return "Division by zero is not allowed"
    return a / b


def calculator():
    result = float(input("Enter first number: "))

    while True:
        op = input("Enter operator (+, -, *, /) or '=' to finish: ")

        if op == "=":
            print("Final Result =", result)
            break

        num = float(input("Enter next number: "))

        if op == "+":
            result = add(result, num)
        elif op == "-":
            result = sub(result, num)
        elif op == "*":
            result = mul(result, num)
        elif op == "/":
            result = div(result, num)
        else:
            print("Invalid Operator")
            continue

        print("Current Result =", result)


calculator()
```

```python
def average(*args):
    total = 0

    for i in args:
        total += i

    return total / len(args)

print(average(10, 20, 30, 40))
```

```python
def my_filter(func, iterable):
    result = []

    for item in iterable:
        if func(item):       # Agar True return hua
            result.append(item)

    return result


nums = [1, 2, 3, 4, 5, 6]

print(my_filter(lambda x: x % 2 == 0, nums))
```

```python
def logger(func):

    def wrapper():
        print("Function Started")
        func()
        print("Function Ended")

    return wrapper


@logger
def greet():
    print("Hello")


greet()
```

```python
memo = {}

def fibonacci(n):
    if n in memo:
        return memo[n]

    if n <= 1:
        return n

    memo[n] = fibonacci(n - 1) + fibonacci(n - 2)
    return memo[n]


print(fibonacci(10))
```

```python
counter = 0

def count_calls(func):
    def wrapper():
        global counter
        counter += 1
        return func()

    return wrapper


@count_calls
def hello():
    print("Hello")


hello()
hello()
hello()

print(counter)
```
