```python
class Student:
    def __init__(self,name,age,marks):
        self.name = name
        self.age = age
        self.marks = marks

    def display(self):
        print(f'Name : {self.name}')
        print(f'Age : {self.age}')
        print(f'Marks : {self.marks}')
        print('-'*20)

student1 = Student('Rishabh',20,92)
student2 = Student('Kunal',20,22)
student3 = Student('Amit',20,52)

student1.display()
student2.display()
student3.display()
        
```

```python
class Car:
    def start(self):
        print('Car Started')
    def display(self):
        print('This is a car')
    def stop(self):
        print('car stopped')

car1 = Car()
car1.start()
car1.display()
car1.stop()
```

```python
class BankAccount:
    def __init__(self,balance):
        self.balance = balance

    def deposit(self,amount):
        self.balance += amount
        print(f"Rs{amount} deposit successfully.")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f'RS{amount} withdrawn successfully.')
        else:
            print('Insufficient Balance!')

    def check_balance(self):
        print(f'Current Balance: RS{self.balance}')

account = BankAccount(5000)

account.check_balance()
account.deposit(2000)
account.withdraw(3000)
account.check_balance()
```

```python
class Employee:
    def __init__(self,name,monthly_salary):
        self.name = name
        self.monthly_salary = monthly_salary

    def yearly_salary(self):
        print(self.monthly_salary*12)

e1 = Employee('Rishabh',50000)
e1.yearly_salary()
```

```python
class Rectangle:
    def __init__(self,length,width):
        self.length = length
        self.width = width

    def area(self):
        print(f'Area of Rectangle: {self.length*self.width}')

    def perimeter(self):
        print(f"Perimeter of Rectangle: {2*(self.length+self.width)}")

r1 = Rectangle(5,6)
r1.area()
r1.perimeter()
```

```python
class Book:
    def __init__(self,title):
        self.title = title
        self.is_issued = False

    def issue(self):
        self.is_issued=True
        print(f"{self.title} has been issued.")

    def return_book(self):
        self.is_issued = False
        print(f"{self.title} has been returned.")

book1 = Book('Python')

book1.issue()
book1.return_book()
```

```python
import math

class Circle:

    def __init__(self, radius):

        if radius <= 0:
            raise ValueError("Radius must be positive.")

        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def circumference(self):
        return 2 * math.pi * self.radius

    def display(self):
        print(f"Radius: {self.radius}")


c1 = Circle(5)
c2 = Circle(10)
c1.display()
print(f"Area: {c1.area():.2f}")
print(f"Circumference: {c1.circumference():.2f}")

c2.display()
print(f"Area: {c2.area():.2f}")
print(f"Circumference: {c2.circumference():.2f}")
```
