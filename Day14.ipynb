```python
class Student:

    # Class Variable
    college = "ABC College"

    def __init__(self, name, age, marks):
        # Instance Variables
        self.name = name
        self.age = age
        self.marks = marks


# Creating 3 objects
s1 = Student("Rishabh", 20, 85)
s2 = Student("Aman", 21, 78)
s3 = Student("Rahul", 19, 92)


# 1. Every student's details
print(s1.name, s1.age, s1.marks)
print(s2.name, s2.age, s2.marks)
print(s3.name, s3.age, s3.marks)


# 2. College
print(s1.college)
print(s2.college)
print(s3.college)


# 3. Change one student's name
s1.name = "Rohit"

print(s1.name)
print(s2.name)
print(s3.name)


# 4. Change class variable
Student.college = "XYZ College"

print(s1.college)
print(s2.college)
print(s3.college)
```

```python
class Employee:
    company = 'TechCrop'

    @classmethod
    def change_company(cls,new_company):
        cls.company = new_company

e1 = Employee()
e2 = Employee()
e3 = Employee()

print(e1.company)
print(e2.company)
print(e3.company)

Employee.change_company("Google")

print(e1.company)
print(e2.company)
print(e3.company)
```

```python
class MathUtils:

    @staticmethod
    def is_even(n):
        return n % 2 == 0

    @staticmethod
    def is_prime(n):
        if n < 2:
            return False

        for i in range(2, int(n ** 0.5) + 1):
            if n % i == 0:
                return False

        return True


print(MathUtils.is_even(10))
print(MathUtils.is_even(7))

print(MathUtils.is_prime(17))
print(MathUtils.is_prime(20))
```

```python
class Student:

    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks

    @classmethod
    def from_string(cls, data):
        name, age, marks = data.split(",")

        return cls(name, int(age), int(marks))


data = "Rishabh,20,85"

student1 = Student.from_string(data)

print("Name:", student1.name)
print("Age:", student1.age)
print("Marks:", student1.marks)
```

```python
class Employee:
    company = "Google"
    employee_count = 0

    def __init__(self,name,salary):
        self.name = name
        self.salary = salary

        Employee.employee_count += 1

    @classmethod
    def change_company(cls,new_company):
        cls.company = new_company

    @staticmethod
    def is_valid_salary(salary):
        return salary > 0

    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)
        print("Company:", self.company)
            

e1 = Employee('Rishabh',10000)
e2 = Employee('Tushar',20000)

Employee.change_company("Microsoft")

e1.display()
e2.display()

print("Employee count:", Employee.employee_count)
print(Employee.is_valid_salary(5000))
```
