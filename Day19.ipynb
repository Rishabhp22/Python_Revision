```python
df['Marks']=df['Marks'].apply(lambda X:X+5)
df
```

```python
df['Result'] = df['Marks'].apply(lambda X: 'Pass' if X >= 40 else 'Fail')
df
```

```python
df.loc[df['Name'] == 'C', 'Marks'] -=70
```

```python
df
```

```python
df['Grade'] = df['Marks'].apply(
    lambda x: 'A' if x >= 90
    else 'B' if x >= 75
    else 'C' if x >= 60
    else 'D' if x >= 40
    else 'F'
)
```

```python
df
```

```python
df['Remark'] = df['Grade'].apply(
    lambda x: 'Excellent' if x == 'A'
    else 'Good' if x == 'B'
    else 'Average' if x == 'C'
    else 'Needs Improvement' if x == 'D'
    else 'Fail'
)
```

```python
df
```

```python
df['Age'] = [20,21,18,19,22,20]
```

```python
df['Status'] = df.apply(lambda row: 'Eligible' if (row['Marks'] >= 80) & (row['Age'] <= 20) else 'Not Eligible',axis=1)
df
```

```python
df.insert(2, "Age", df.pop("Age"))
```

```python
df
```

```python
df['Extra Marks'] = [10,5,2,5,3,8]
```

```python
df["Marks"] += df["Extra Marks"]
df
```

```python
df['Category'] = df.apply(
    lambda row: 'High Performer' if (row['Marks'] >= 80) & (row['Age'] <= 20) else 'Good Performer'
    if (row['Marks'] >= 60) & (row['Age'] <= 20) else 'Average',axis=1
)
df
```
