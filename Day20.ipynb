```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E", "F"],
    "Class": ["A", "A", "B", "B", "A", "B"],
    "Gender": ["M", "F", "M", "F", "M", "F"],
    "Marks": [90, 77, 26, 73, 100, 83]
})
```

```python
df.groupby("Class")["Marks"].mean()
```

```python
pd.pivot_table(
    df,
    values="Marks",
    index="Class",
    aggfunc="mean"
)
```

```python
pd.pivot_table(
    df,
    values='Marks',
    index='Class',
    aggfunc='mean'
)
```

```python
pd.pivot_table(
    df,
    values='Marks',
    index='Class',
    aggfunc=['mean','max']
)
```

```python
pd.pivot_table(
    df,
    values="Marks",
    index="Class",
    columns="Gender",
    aggfunc="mean"
)
```

```python
pd.pivot_table(
    df,
    values='Marks',
    index='Class',
    aggfunc='sum'
)
```

```python
pd.pivot_table(
    df,
    values='Marks',
    index='Class',
    columns='Gender',
    aggfunc=['mean','max','count']
)
```

```python
df.groupby(["Class", "Gender"])["Marks"].agg(["mean", "max", "count"])
```

```python
df1 = pd.DataFrame({
    "ID": [1, 2, 3, 4],
    "Name": ["A", "B", "C", "D"],
    "Class": ["A", "A", "B", "B"]
})

df2 = pd.DataFrame({
    "ID": [3, 4, 5, 6],
    "Marks": [70, 60, 88, 92],
    "Grade": ["C", "D", "B", "A"]
})
```

```python
pd.merge(
    df1,df2,
    on='ID',
    how='inner'
)
```

```python
pd.merge(
    df1,df2,
    on='ID',
    how='left'
)
```

```python
merge_df = pd.merge(
    df1,df2,
    on='ID'
)
```

```python
merge_df[['Name','Class','Marks','Grade']]
```

```python
merge_df[merge_df['Marks'] >= 80]
```

```python
merge_df.sort_values('Marks', ascending=False)
```

```python
merge_df = pd.merge(df1, df2, on='ID')

merge_df.groupby('Class')['Marks'].mean()
```

```python
merge_df['Result'] = merge_df['Marks'].apply(lambda x: 'Pass' if x >= 40 else 'Fail')
```

```python
merge_df
```

```python
merge_df[merge_df['Class'] == 'B']['Marks'].agg(
    ['max', 'min', 'mean']
)
```

```python
merge_df[merge_df['Class'] == 'B']['Marks'].agg(
    highest_marks='max',
    lowest_marks='min',
    average_marks='mean'
)
```

```python
df1 = pd.DataFrame({
    "ID": [1, 2, 3, 4, 5],
    "Name": ["A", "B", "C", "D", "E"],
    "Class": ["A", "A", "B", "B", "A"]
})

df2 = pd.DataFrame({
    "ID": [1, 2, 3, 4, 5],
    "Marks": [90, 65, 82, 35, 76],
    "Age": [20, 21, 19, 20, 22]
})
```

```python
merge = pd.merge(
    df1,df2,
    on='ID'
)
```

```python
merge[(merge['Class']=='B') & (merge['Age']<=20)].sort_values('Marks',ascending=False)[['Name','Marks','Class','Age']]
```
