```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
models = [
    'Linear Regression',
    'Decision Tree',
    'Random Forest',
    'Support Vector Machine'
]

accuracy = [78, 82, 89, 85]
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    models,
    accuracy,
    color='skyblue',
    edgecolor='black',
    width=0.7
)

plt.title('Model Accuracy Comparison',fontsize=16,fontweight='bold')
plt.xlabel('Models',fontsize=12)
plt.ylabel('Accuracy (%)',fontsize=12)
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)

for i,value in enumerate(accuracy):
    plt.text(i,value +1, f'{value}%', ha='center', fontweight='bold')

plt.tight_layout()

plt.show()
```

```python
models = ['Decision Tree', 'Random Forest', 'SVM']
train_accuracy = [88, 95, 91]
test_accuracy = [82, 89, 85]
```

```python
import numpy as np

width = 0.35
x = np.arange(len(models))
plt.figure(figsize=(10,6))
plt.bar(
    x - width/2,
    train_accuracy,
    width=width,
    color='blue',
    edgecolor='black',
    label='Training Accuracy'
)

plt.bar(
    x + width/2,
    test_accuracy,
    width=width,
    color='orange',
    edgecolor='black',
    label='Testing Accuracy'
)

plt.title('Traing vs Testing accuracy', fontsize=20, fontweight='bold')
plt.xlabel('Models', fontsize=17)
plt.ylabel('Accuracy (%)', fontsize=17)
plt.xticks(x, models)
plt.grid(axis='y', alpha=0.3)

for i, value in enumerate(train_accuracy):
    plt.text(
        x[i] - width/2,
        value + 1,
        f'{value}%',
        ha='center',
        fontweight='bold'
    )

for i, value in enumerate(test_accuracy):
    plt.text(
        x[i] + width/2,
        value + 1,
        f'{value}%',
        ha='center',
        fontweight='bold'
    )

plt.legend()
plt.tight_layout()
plt.show()
```

```python
models = ['Decision Tree', 'Random Forest', 'SVM', 'Logistic Regression']

train_accuracy = [88, 95, 91, 86]
test_accuracy = [82, 89, 85, 83]
```

```python
width=0.35
x = np.arange(len(models))
plt.figure(figsize=(12,6))

plt.bar(
    x - width/2,
    train_accuracy,
    width=width,
    color='blue',
    edgecolor='black',
    label='Train Accuracy'
)

plt.bar(
    x + width/2,
    test_accuracy,
    width=width,
    color='orange',
    edgecolor='black',
    label='Test Accuracy'
)

plt.title('Training vs Testing Model Accuracy', fontsize=18, fontweight='bold')
plt.xlabel('Models', fontsize=12)
plt.ylabel('Accuracy (%)', fontsize=12)
plt.xticks(x, models)
plt.grid(axis='y', alpha=0.3)
plt.legend()

for i,value in enumerate(train_accuracy):
    plt.text(
        x[i] - width/2,
        value + 1,
        f"{value}%",
        ha = 'center',
        fontweight='bold'
    )


for i,value in enumerate(test_accuracy):
    plt.text(
        x[i] + width/2,
        value + 1,
        f"{value}%",
        ha = 'center',
        fontweight='bold'
    )

plt.tight_layout()
plt.show()
```
