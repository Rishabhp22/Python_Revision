```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
experience = [1, 2, 3, 4, 5, 6, 7]
salary = [25, 28, 32, 38, 45, 52, 60]
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    experience,
    salary,
    color='green',
    s=100,
    alpha=0.7,
    marker='o',
    edgecolor='black',
    linewidth=1
)

plt.xlabel('Experience (Years)')
plt.ylabel('Salary (LPA)')
plt.title('Experience vs Salary')
plt.grid()
plt.show()
```

```python
boys_hours = [2, 3, 4, 5, 6]
boys_marks = [45, 52, 60, 68, 75]

girls_hours = [2, 3, 4, 5, 6]
girls_marks = [48, 56, 65, 72, 82]
```

```python
plt.scatter(
    boys_hours,
    boys_marks,
    label='Boys'
)

plt.scatter(
    girls_hours,
    girls_marks,
    label='Girls'
)

plt.legend()
plt.show()
```

```python
study_hours_A = [1, 2, 3, 4, 5]
marks_A = [40, 48, 55, 65, 72]

study_hours_B = [1, 2, 3, 4, 5]
marks_B = [45, 53, 62, 70, 82]
```

```python
plt.scatter(
    study_hours_A,
    marks_A,
    marker='o',
    color='pink',
    edgecolor='black',
    linewidth=1.5,
    s=100,
    alpha=0.7,
    label='Marks A'
)

plt.scatter(
    study_hours_B,
    marks_B,
    marker='^',
    color='blue',
    edgecolor='black',
    linewidth=1.5,
    s=100,
    alpha=0.7,
    label='Marks B'
)

plt.title('Study Hours vs Marks')
plt.xlabel('Marks')
plt.ylabel('Hours')
plt.legend()
plt.grid()
plt.show()
```
