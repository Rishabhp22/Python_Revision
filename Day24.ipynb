```python
marks = [12, 18, 25, 32, 38, 41, 45, 49, 52, 57,
         61, 65, 69, 72, 76, 81, 85, 89, 94, 98]
```

```python
bins = [40,50,60,70,80,90,100]
plt.hist(
    marks,
    bins=bins,
    histtype='bar',
    color='pink',
    edgecolor='black'
)
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.title('Marks Frequency')
plt.grid(axis='y',alpha=0.8)
plt.show()
```

```python
bins = [40,50,60,70,80,90,100]
plt.hist(
    marks,
    bins=bins,
    histtype='barstacked',
    color='pink',
    edgecolor='black'
)
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.title('Marks Frequency')
plt.grid(axis='y',alpha=0.8)
plt.show()
```

```python
bins = [40,50,60,70,80,90,100]
plt.hist(
    marks,
    bins=bins,
    histtype='step',
    color='pink',
    edgecolor='black'
)
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.title('Marks Frequency')
plt.grid(axis='y',alpha=0.8)
plt.show()
```

```python
bins = [40,50,60,70,80,90,100]
plt.hist(
    marks,
    bins=bins,
    histtype='stepfilled',
    color='pink',
    edgecolor='black'
)
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.title('Marks Frequency')
plt.grid(axis='y',alpha=0.8)
plt.show()
```

```python
boys = [40, 45, 50, 55, 60, 65]
girls = [42, 48, 52, 58, 62, 68]

plt.hist(
    [boys, girls],
    bins=5,
    histtype='barstacked',
    label=['Boys', 'Girls']
)

plt.legend()
plt.show()
```

```python
boys = [45, 50, 52, 55, 60, 62, 65, 70, 72, 75]
girls = [48, 51, 54, 58, 63, 66, 68, 73, 78, 82]
```

```python
plt.hist(boys, bins=5, alpha=0.5, label='Boys')
plt.hist(girls, bins=5, alpha=0.5, label='Girls')

plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.title('Boys vs Girls Marks Distribution')
plt.legend()
plt.show()
```

```python
plt.hist(boys, bins=5, alpha=0.5, label='Boys',histtype='barstacked')
plt.hist(girls, bins=5, alpha=0.5, label='Girls')

plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.title('Boys vs Girls Marks Distribution')
plt.legend()
plt.show()
```
