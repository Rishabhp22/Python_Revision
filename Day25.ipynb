```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
plans = ['Basic', 'Standard', 'Premium']
customers = [500, 300, 200]

plt.pie(customers, labels=plans, autopct='%1.1f%%')
plt.show()
```

```python
plt.pie(
    customers,
    labels=plans,
    autopct='%1.1f%%',
    startangle=90
)
plt.show()
```

```python
plans = ['Basic', 'Standard', 'Premium']
customers = [500, 300, 200]

plt.pie(
    customers,
    labels=plans,
    labeldistance=1.1,
    autopct='%1.1f%%',
    pctdistance=0.5,
    explode=[0, 0.1, 0],
    shadow=True,
    colors=['red','pink','magenta'],
    wedgeprops={
        'edgecolor':'black',
        'linewidth':2
    },
    counterclock=True,
    radius=2,
    startangle=90,
    rotatelabels=90,
    textprops={
        'fontsize': 16,
        'fontweight':'bold'
    }
)

plt.show()
```

```python
models = ['Logistic Regression', 'Decision Tree', 'Random Forest', 'SVM']
predictions = [250, 150, 400, 200]
```

```python
wedges, texts, autotexts = plt.pie(
    predictions,
    labels=models,
    autopct='%1.1f%%',
    explode=[0,0,0.08,0],
    shadow=True,
    colors=['skyblue','red','purple','brown'],
    wedgeprops={
        'edgecolor':'black',
        'linewidth':1
    },
    pctdistance=0.65,
    radius=1.1,
    startangle=90
)

for text in texts:
    text.set_fontsize(13)
    text.set_fontweight('bold')

for autotext in autotexts:
    autotext.set_fontsize(11)
    autotext.set_fontweight('bold')

plt.title('Models Predictions (pie chart)', fontweight='bold', fontsize=15)
plt.tight_layout()
plt.show()
```
