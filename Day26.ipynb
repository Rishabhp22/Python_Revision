```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.subplot(1, 3, 1)
plt.plot(x, y)
plt.title('Line Plot')

plt.subplot(1, 3, 2)
plt.bar(x, y)
plt.title('Bar Chart')

plt.subplot(1,3,3)
plt.scatter(x,
            y,
            c=x,
            cmap='viridis')
plt.title('Scatter Plot')
plt.colorbar()


plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]
```

```python
fig, ax =plt.subplots(2,2, figsize=(10,5))

ax[0,0].plot(x,y)
ax[0,0].set_title('Line Plot')
ax[0,0].set_xlabel('X')
ax[0,0].set_ylabel('Y')

ax[0,1].bar(x,y)
ax[0,1].set_title('Bar Plot')
ax[0,1].grid(axis='y')

ax[1,0].scatter(x,y)
ax[1,0].set_title('Scatter Plot')
ax[1,0].set_xlabel('X')
ax[1,0].set_ylabel('Y')

ax[1,1].hist(y, bins=5)
ax[1,1].set_title('Histogram Plot')

plt.tight_layout()
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
sales = [10, 20, 15, 25, 30]
profit = [5, 12, 8, 18, 22]
```

```python
fig, ax = plt.subplots(1,2, figsize=(10,4))
fig.suptitle('Graph')

ax[0].plot(x,sales)
ax[0].set_title('Sales')
ax[0].set_xlabel('Days')
ax[0].set_ylabel('Sales')
ax[0].set_ylim(0,35)

ax[1].bar(x,profit)
ax[1].set_xlabel('Days')
ax[1].set_title('Profit')
ax[1].set_ylabel('Profit')
ax[1].grid(axis='y')

plt.tight_layout()
plt.show()

```
