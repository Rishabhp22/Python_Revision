```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y, marker='o')

plt.annotate(
    'Highest',
    xy=(4, 30),
    xytext=(3, 35),
    arrowprops=dict(arrowstyle='->')
)
plt.tick_params(axis='both', labelsize=12)
plt.axhline(y=20, color='red', linestyle='--')
plt.fill_between(x, y)

plt.show()
```
